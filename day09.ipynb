# Task1: المتوسط يكذب احيانا

```python
import numpy as np
```

```python
salaries = np.array([6000, 6500, 7000, 7200, 7500, 8000, 95000])
```

# سؤال: صاحب الشركة نشر اعلان يقول ان متوسط الرواتب في الشركة هو 19,600

أعتقد أن الإعلان مضلل، لأن راتب المدير
التنفيذي مرتفع جدًا مقارنة برواتب باقي الموظفين، وهذا رفع المتوسط الحسابي

الوسيط 7200 ريال يعكس الرواتب بشكل أفضل.

```python
# حساب المتوسط والوسيط

mean_salary = np.mean(salaries)
median_salary = np.median(salaries)

print("Mean:", mean_salary)
print("Median:", median_salary)
```

```python
# حساب Q1 , Q3 , IQR

Q1 = np.percentile(salaries, 25)
Q3 = np.percentile(salaries, 75)

IQR = Q3 - Q1

lower = Q1 - (1.5 * IQR)
upper = Q3 + (1.5 * IQR)

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

```python
# تحديد Outlier
#غالبا القيمة الشاذة هي راتب المدير لانها اعلى من "upper bound"

outliers = salaries[(salaries < lower) | (salaries > upper)]

print("Outliers:", outliers)
```

```python
# الحساب بدون راتب المدير

salaries_without_manager = salaries[:-1]

mean_without = np.mean(salaries_without_manager)
median_without = np.median(salaries_without_manager)

print("Mean without manager:", mean_without)
print("Median without manager:", median_without)
```

بعد حذف راتب المدير انخفض المتوسط بشكل كبير بينما تغير الوسيط بشكل بسيط

 نستنتج أن المتوسط يتأثر بالقيم الشاذة أكثر من الوسيط

# امثلة outlier

1. ارتفاع عدد الزوار في مدينة الرياض خلال الاجازه او موسم معين < هذي تعتبر قيمة شاذة لكن حقيقة ومهمه ولايجب حذفها

2. سرعة الرياح اثناء عاصفة قوية < قد تكون اعلى بكثير من من الارقام الاعتيادية لكنها حقيقة ومهمه لدراسة الطقس ولا يجب حذفها

3. تسجيل درجة حرارة 500 بدل من 50 < هنا قيمة شاذة خاطئة بسبب غلط في ادخال البيانات

# ماهو z-score?

هو مقياس يوضح لنا كم تبعد قيمة معينة عن المتوسط، وهل هي أعلى أو أقل منه، باستخدام الانحراف المعياري.

مثال: حصلوا اغلب الطالبات على درجات متقاربة في الاختبار ولكن في طالبة حصلت على درجة مختلفة وبعيده جدا عن بقية الدرجات

z-score يقدر يحدد مدى  اختلاف درجتها عن المتوسط

```python
import numpy as np

data = np.genfromtxt("insurance.csv", delimiter=",", skip_header=1, usecols=(0, 2, 3, 6))

np.set_printoptions(suppress=True)

print("Shape:", data.shape)
print("First 3 rows:")
print(data[:3])
```

# Task3: هل هذه اخطاء؟

```python
# حساب IQR & Outliers

columns = [charges, bmi]
names = ["Charges", "BMI"]

for i in range(2):
    Q1 = np.percentile(columns[i], 25)
    Q3 = np.percentile(columns[i], 75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    outliers = columns[i][(columns[i] < lower) | (columns[i] > upper)]

    percentage = (len(outliers) / len(columns[i])) * 100

    print(names[i])
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
    print("Outliers:", len(outliers))
    print("Percentage:", percentage)
```

```python
# طباعة اعلى 5 قيم في كل عمود

print("Top 5 Charges:")
print(np.sort(charges)[-5:])

print("Top 5 BMI:")
print(np.sort(bmi)[-5:])
```

```python
# حساب z-score

z_charges = (charges - np.mean(charges)) / np.std(charges)
z_bmi = (bmi - np.mean(bmi)) / np.std(bmi)

print("Charges:", np.sum(np.abs(z_charges) > 3))
print("BMI:", np.sum(np.abs(z_bmi) > 3))
```

```python
# مقارنة متوسط التكاليف

Q1 = np.percentile(charges, 25)
Q3 = np.percentile(charges, 75)

IQR = Q3 - Q1

lower = Q1 - (1.5 * IQR)
upper = Q3 + (1.5 * IQR)

normal = charges[(charges >= lower) & (charges <= upper)]

mean_with = np.mean(charges)
mean_without = np.mean(normal)

print("Mean with Outliers:", mean_with)
print("Mean without Outliers:", mean_without)
```

# Task 4: كل الأعمدة على مقياس واحد

```python
# Min-Max Normalization

age_norm = (age - age.min()) / (age.max() - age.min())
bmi_norm = (bmi - bmi.min()) / (bmi.max() - bmi.min())
charges_norm = (charges - charges.min()) / (charges.max() - charges.min())

print(age_norm.min(), age_norm.max())
print(bmi_norm.min(), bmi_norm.max())
print(charges_norm.min(), charges_norm.max())
```

```python
# بيانات العميل قبل وبعد التحويل

print("Before:")
print(age[0], bmi[0], charges[0])

print("After:")
print(age_norm[0], bmi_norm[0], charges_norm[0])
```

```python
# Standardization (z-score)

charges_std = (charges - charges.mean()) / charges.std()

print("Mean:", charges_std.mean())
print("Std:", charges_std.std())
```

```python
# تحويل بيانات العميل الجديد باستخدام min و max من البيانات الأصلية


new_age = (40 - age.min()) / (age.max() - age.min())
new_charges = (20000 - charges.min()) / (charges.max() - charges.min())

print("Age:", new_age)
print("Charges:", new_charges)
```

# Task5: صيد الكواكب

```python
# انشاء البيانات

rng = np.random.default_rng(7)

brightness = 1000 + rng.normal(0, 1, 240)

for start in [30, 102, 174]:
    brightness[start:start + 5] -= 20
```

```python
# حساب المتوسط والوسيط

mean_brightness = np.mean(brightness)
median_brightness = np.median(brightness)

print("Mean:", mean_brightness)
print("Median:", median_brightness)
```

```python
# حساب z-score

z_score = (brightness - mean_brightness) / np.std(brightness)

mask = z_score < -3

print("Number of readings:", np.sum(mask))
print("Hours:", np.arange(240)[mask])
```

```python
# تحديد بداية كل عبور

changes = np.diff(mask.astype(int))

starts = np.where(changes == 1)[0] + 1

print("Transit start hours:", starts)
```

```python
# حساب الزمن بي العبورين

time_between = np.diff(starts)

days = time_between / 24

print("Hours between transits:", time_between)
print("Days between transits:", days)
```

```python
# حساب مقدار انخفاض السطوع

normal_brightness = 1000

decrease = normal_brightness - np.mean(brightness[mask])

percentage = (decrease / normal_brightness) * 100

print("Brightness decrease:", decrease)
print("Percentage decrease:", percentage)
```
